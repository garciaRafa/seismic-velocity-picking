# v3 — Particle Swarm Optimization: Search Visualization

How PSO (with or without Lévy flights) moves through the search space on
the synthetic FD model, **with and without free-surface multiples**: the
objective along the run, the swarm (best, number of active slots and
spread), the number of picks, snapshots, where the evaluations were
spent, and an animation.

PSO moves **particles**: real vectors of fixed length pulled toward their
own best position and toward the best position of the swarm. A picking
solution has a variable number of picks, so each particle holds 11
**slots** (time, velocity, presence) and a slot is a pick when its
presence is above 0.5; the active slots are sorted in time and the
largest subset that respects the constraints is kept. When the swarm
best stops improving for a while, every particle except the best one is
re-drawn at random. Optionally (`LEVY = True`), particles take Lévy-flight
steps: mostly small, occasionally very large.

**Why PSO is expected to struggle here.** The semblance panel is almost
zero everywhere except on small blobs (about ±20 ms and ±2% in velocity),
so a particle moving continuously gets no signal until it lands on one.
And slot i of one particle is not necessarily the same event as slot i of
the swarm best (one has found the top of A, the other has not): pulling
one toward the other can move a correct pick onto a different event. The
other algorithms work pick by pick (insert, remove, shift one pick) and
do not have this problem. This notebook shows it.

Budget: only objective evaluations count (one per particle per
iteration), as for the other algorithms.

Formulation: the current baseline, `configs/batch_fd_multicdp_v3_ma_pso.json`
(v3, robust semblance, 200 ms separation, minimum velocity gradient
200 m/s per s, picks up to 2150 ms, reference with the equivalent maxima of
each event, 100,000 evaluations). Set `LEVY = True` in section 2 for the
Lévy-flight variant (figures saved with `_levy`).

**Execution order:**
1. Setup

3. The data: shot record, CDP gather and semblance panel
4. Load the cases
5. Run on both cases, with the trajectory recorded
6. Objective and number of picks along the run

## 2. Parameters

```python
ALG = 'Particle Swarm Optimization'
TAG = 'pso'

# --- Cases: the synthetic FD model with and without free-surface multiples ---
CASES = ['fd_layered_cdp2500_fs', 'fd_layered_cdp2500_nofs']
LABEL = {'fd_layered_cdp2500_fs':   'with free surface (multiples)',
         'fd_layered_cdp2500_nofs': 'without free surface'}

# --- Formulation: same as configs/batch_fd_multicdp_v3_ma_pso.json (current baseline) ---
VEL_MIN      = 1400.0
VEL_MAX      = 5000.0
VEL_STEP     = 20.0          # panel used for tau and for the figures
MIN_SEP_MS   = 200.0
T_MAX_MS     = 2150.0        # latest pick: the side-boundary reflection arrives after it
TAU_FACTOR   = 10.0
P_INSERT     = 0.1
P_REMOVE     = 0.1
SEMBLANCE    = ROBUST_SEMBLANCE   # stretch mute 0.5, energy floor 1%, min fold 10
MIN_VEL_GRADIENT = 200.0     # m/s per second between consecutive picks; 0 = monotonic
REFERENCE_ALT_RATIO = 0.9    # a pick on an equivalent maximum of an event counts (None: peak only)
CACHE_SEMBLANCE = True       # same results, faster

# --- Budget and seed (the same for both cases) ---
MAX_EVALS    = 100000
SEED         = 42
RECORD_EVERY = 1             # one record per swarm iteration (every particle moved once)


# --- Figures ---
V_SHOW       = 3000.0        # velocity axis of the panels (m/s)

# --- Particle Swarm Optimization ---
LEVY           = False       # True: Lévy-flight steps (ParticleSwarmLevy in the batch configs)
PSO_SWARM      = 30          # particles
PSO_W          = 0.72        # inertia
PSO_C1         = 1.49        # cognitive coefficient (pull toward the particle's own best)
PSO_C2         = 1.49        # social coefficient (pull toward the swarm's best)
PSO_VMAX       = 0.2         # max speed per coordinate, fraction of its range
PSO_RESTART    = 50          # iterations without a new swarm best -> re-draw all but the best
PSO_P_LEVY     = 0.2         # (LEVY) probability of a Lévy step per particle and iteration
PSO_LEVY_SCALE = 0.01        # (LEVY) step size, fraction of each coordinate's range
PSO_LEVY_BETA  = 1.5         # (LEVY) index of the Lévy distribution

if LEVY:
    ALG, TAG = 'PSO with Lévy flights', 'pso_levy'

# Output names: runs with a minimum gradient are saved apart
if MIN_VEL_GRADIENT > 0:
    TAG = f'{TAG}_g{MIN_VEL_GRADIENT:g}'
```

## 3. The data: shot record, CDP gather and semblance panel

The two cases side by side, before any picking: a shot record (all
receivers of the line; the red line marks the source), the CDP gather the
algorithm works on, and its semblance panel (same robust semblance as the
objective). With the free surface, the surface multiples appear in all
three views; without it, only the primaries (and weaker diffractions and
interbed energy) remain.

The shot records are in `data/synthetic_fd/` (not versioned); if they are
missing, run `python scripts/model_fd.py`.

```python
SHOT = 1                  # shot shown: 0 = near offset, 1 = middle, 2 = far offset

vels_show = np.arange(VEL_MIN, V_SHOW + VEL_STEP, VEL_STEP, dtype=np.float32)
fig, axes = plt.subplots(2, 3, figsize=(18, 13), gridspec_kw={'width_ratios': [1.6, 1, 1]})
for row, name in enumerate(CASES):
    c  = load_case(os.path.join(project_root, 'cases', f'{name}.npz'))
    tg = c['t_grid']

    # Shot record (time gain for display only)
    shot_path = os.path.join(project_root, 'data', 'synthetic_fd', f'{name}_shots.npz')
    ax = axes[row, 0]
    if os.path.exists(shot_path):
        s  = np.load(shot_path)
        li = s['line_idx']; rx = s['rec_x'][li]
        rec = s['shots'][SHOT][li] * (s['t_ms'] / s['t_ms'][-1])[None, :]
        clip = np.percentile(np.abs(rec), 99)
        ax.imshow(rec.T, aspect='auto', cmap='gray', vmin=-clip, vmax=clip,
                  extent=[rx[0], rx[-1], tg[-1], tg[0]])
        ax.axvline(s['shot_x'][SHOT], color='red', lw=0.8)
        ax.set_title(f"Shot at {s['shot_x'][SHOT]:.0f} m — {LABEL[name]}")
    else:
        ax.text(0.5, 0.5, 'shot records not found: run scripts/model_fd.py',
                ha='center', va='center', transform=ax.transAxes)
    ax.set_ylabel('Time (ms)')

    # CDP gather (as the algorithm sees it: direct-wave mute + time gain)
    g = c['gather']; clip = np.percentile(np.abs(g), 99)
    axes[row, 1].imshow(g.T, aspect='auto', cmap='gray', vmin=-clip, vmax=clip,
                        extent=[c['offsets'][0], c['offsets'][-1], tg[-1], tg[0]])
    axes[row, 1].set_title(f"CDP gather at {c['meta']['model']['cdp_x_m']:.0f} m")

    # Semblance panel, no picks
    p = compute_semblance(g, c['offsets'], vels_show, c['dt_ms'], **SEMBLANCE)
    axes[row, 2].imshow(p.T, aspect='auto', cmap='jet', vmin=0, vmax=1,
                        extent=[VEL_MIN, V_SHOW, tg[-1], tg[0]])
    axes[row, 2].set_title('Semblance (robust)')
for ax, lab in zip(axes[1], ('Receiver position (m)', 'Offset (m)', 'Velocity (m/s)')):
    ax.set_xlabel(lab)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'search_data_fs_nofs.png'), dpi=130, bbox_inches='tight')
plt.show()
```

## 4. Load the cases

For each case: semblance panel and tau with the same options as the
objective, and the reference picks (semblance peak of each expected
event) for the figures — `+` white = primary, `+` magenta = surface
multiple. The algorithm never sees the reference.

```python
data = {}
for name in CASES:
    case = load_case(os.path.join(project_root, 'cases', f'{name}.npz'))
    velocities = np.arange(VEL_MIN, VEL_MAX + VEL_STEP, VEL_STEP, dtype=np.float32)
    panel = compute_semblance(case['gather'], case['offsets'], velocities, case['dt_ms'], **SEMBLANCE)
    data[name] = {
        'case': case, 'gather': case['gather'], 'offsets': case['offsets'],
        'dt': case['dt_ms'], 't_grid': case['t_grid'], 'panel': panel,
        'tau': robust_threshold(panel, factor=TAU_FACTOR),
        'extent': [VEL_MIN, VEL_MAX, case['t_grid'][-1], case['t_grid'][0]],
        'reference': get_reference(case, alt_ratio=REFERENCE_ALT_RATIO, **SEMBLANCE),
    }
    print(f"{name:26s} gather {case['gather'].shape}   tau = {data[name]['tau']:.3f}")
```

## 5. Run on both cases, with the trajectory recorded

`record_trajectory=True` stores the current solution along the run. It
only reads the search state, so each run is identical to the one without
recording (same seed, same result). Same seed and budget on both cases.

```python
def make_optimizer(d):
    return ParticleSwarm(
        d['gather'], d['offsets'], d['dt'], vel_min=VEL_MIN, vel_max=VEL_MAX,
        min_sep_ms=MIN_SEP_MS, t_max_ms=T_MAX_MS, objective='threshold', tau=d['tau'],
        p_insert=P_INSERT, p_remove=P_REMOVE, seed=SEED, max_evals=MAX_EVALS,
        record_trajectory=True, record_every=RECORD_EVERY,
        min_vel_gradient=MIN_VEL_GRADIENT, cache_semblance=CACHE_SEMBLANCE, **SEMBLANCE,
        swarm_size=PSO_SWARM, w=PSO_W, c1=PSO_C1, c2=PSO_C2, v_max_frac=PSO_VMAX,
        restart_after=PSO_RESTART, levy=LEVY, p_levy=PSO_P_LEVY,
        levy_scale=PSO_LEVY_SCALE, levy_beta=PSO_LEVY_BETA)

runs = {}
for name in CASES:
    d = data[name]
    opt = make_optimizer(d).run()
    runs[name] = opt
    t_s, v_s = opt.get_result()
    m = match_reference(t_s * d['dt'], v_s, d['reference'])
    print(f"{name:26s} f = {opt.best_score:6.3f}   K = {len(t_s)}   "
          f"on primary {m['n_on_primary']}/{m['n_primaries']}, on multiple {m['n_on_multiple']}, "
          f"F1 {m['f1']:.2f}   ({opt.exec_time_s:.0f} s, {opt.iteration} iterations)")
    print(f"{'':26s} {opt.stats}")
```

## 6. Objective and number of picks along the run

Top: objective of the last particle of each iteration (blue) and best so
far (red); vertical lines are not drawn for PSO restarts here (see
section 9). Bottom: number of picks K of that particle after decoding.
One column per case.

```python
fig, axes = plt.subplots(2, len(CASES), figsize=(8 * len(CASES), 7), sharex=True, sharey='row')
for col, name in enumerate(CASES):
    search_viz.plot_score_trace(runs[name], ax=axes[0, col])
    axes[0, col].set_title(f'{ALG} — {LABEL[name]}')
    search_viz.plot_n_picks(runs[name], ax=axes[1, col])
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, f'search_{TAG}_trace.png'), dpi=150)
plt.show()
```

## 7. Snapshots of the current particle

Six moments of each run, evenly spaced (one row per case). White circles:
the last decoded particle of that iteration; black X: best solution found
up to that moment (when different).

```python
fig, axes = plt.subplots(len(CASES), 6, figsize=(20, 7 * len(CASES)), sharey=True)
for row, name in enumerate(CASES):
    d = data[name]
    search_viz.plot_snapshots(runs[name], d['panel'], d['extent'], d['dt'], n=6,
                              reference=d['reference'], axes=axes[row])
    for ax in axes[row]:
        ax.set_xlim(VEL_MIN, V_SHOW)
    axes[row, 0].set_ylabel(f'{LABEL[name]}\nTime (ms)')
plt.suptitle(f'{ALG} — current solution during the run', y=1.0)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, f'search_{TAG}_snapshots.png'), dpi=110, bbox_inches='tight')
plt.show()
```

## 8. Where the search looked

Every new pick the algorithm evaluated (a pick of an evaluated solution
that was not in the latest child), as a 2D histogram over the panel (gray).
Dense regions are where the evaluations were spent; regions without color
were never tried.

```python
fig, axes = plt.subplots(1, len(CASES), figsize=(8 * len(CASES), 8))
for ax, name in zip(axes, CASES):
    d = data[name]
    search_viz.plot_proposals(runs[name], d['panel'], d['extent'], d['dt'], ax=ax,
                              reference=d['reference'])
    ax.set_title(f'{ALG} — {LABEL[name]}\n{len(runs[name].proposals)} new picks evaluated')
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, f'search_{TAG}_proposals.png'), dpi=130)
plt.show()
```

## 9. Swarm: best, active slots and spread

Top: best objective of the swarm along the run. Middle: mean number of
active slots per particle (presence > 0.5) — the number of picks the
particles try before the repair. Bottom: spread of the swarm (standard
deviation of the slot times and velocities across particles, as a
fraction of their range): it falls as the swarm converges and jumps back
at each restart (vertical lines).

```python
fig, axes = plt.subplots(3, len(CASES), figsize=(8 * len(CASES), 9), sharex=True)
for col, name in enumerate(CASES):
    opt  = runs[name]
    traj = [r for r in opt.trajectory if r['event'] == 'iter']
    ev   = [r['n_evals'] for r in traj]
    axes[0, col].plot(ev, [r['gbest_score'] for r in traj], color='tab:red')
    axes[0, col].set_ylabel('f (swarm best)'); axes[0, col].set_title(LABEL[name])
    axes[1, col].plot(ev, [r['mean_active_slots'] for r in traj], color='tab:blue')
    axes[1, col].set_ylabel('Active slots (mean)')
    axes[2, col].plot(ev, [r['swarm_spread'] for r in traj], color='tab:green')
    axes[2, col].set_ylabel('Swarm spread'); axes[2, col].set_xlabel('Objective evaluations')
    restarts = [r['n_evals'] for a, r in zip(traj, traj[1:]) if r['restarts'] > a['restarts']]
    for ax in axes[:, col]:
        ax.grid(True, alpha=0.3)
        for e in restarts:
            ax.axvline(e, color='gray', lw=0.5, ls='--', alpha=0.6)
    print(f"{name:26s} iterations {opt.iteration}, restarts {opt.stats['restarts']}, "
          f"Lévy steps {opt.stats['levy_steps']}")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, f'search_{TAG}_swarm.png'), dpi=150)
plt.show()
```

## 10. Animation

One GIF per case: the current solution moving over the panel (subsampled
to at most `MAX_FRAMES` frames), saved in `results/figures/`. It can take
a minute per case.

```python
MAX_FRAMES = 200
FPS        = 10
for name in CASES:
    d = data[name]
    gif = os.path.join(FIG_DIR, f'search_{TAG}_{name}.gif')
    search_viz.animate_search(runs[name], d['panel'], d['extent'], d['dt'], gif,
                              max_frames=MAX_FRAMES, fps=FPS, reference=d['reference'],
                              title=f'{ALG} — {LABEL[name]}')
    print(f'Saved: {gif}')
    display(Image(filename=gif))
```

## 11. Summary

**What to look for in PSO:**
- in section 8, the evaluations spread over large empty areas of the
  panel: a particle gets no signal between the blobs, so most of the
  budget is spent where the semblance is zero;
- the swarm best (section 9) improves in a few jumps and then stays flat
  until a restart; the restarts behave like new random starts;
- the picks found are usually the strongest, shallowest events (the
  water bottom, the top of A): finding a deep primary needs a slot to
  land on its blob by chance and the other picks to stay valid;
- compare with the Memetic Algorithm notebook (same cases, budget and
  formulation): the difference comes from the representation — slots of
  a fixed-length vector against picks inserted, removed and shifted one
  at a time;
- with `LEVY = True`, the long Lévy steps make the swarm explore more
  between restarts; check whether that is enough to find more blobs.

**With vs. without free surface.** The same algorithm, seed and budget on
both cases: the differences between the two columns come from the data.
